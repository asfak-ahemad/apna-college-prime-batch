```python
'''B. Select only those rows where: 
    • petal_length > 4.5 
    • species = " Iris-virginica" '''
```

```python
filtered_df = df[(df["petal_length"] > 4.5) & (df["species"] == "Iris-virginica")]
filtered_df
```

```python
df["species"] == "Iris-virginica"
```

```python
'''C. Group by species and compute: 
    •  Average sepal_length 
    •  Maximum petal_width 
    •  Standard deviation of sepal_width'''
```

```python
species_summary = df.groupby("species").agg({
    "sepal_length": "mean",
    "petal_width": "max",
    "sepal_width": "std"
})

species_summary
```

```python
'''D. Create a new column:  "petal_ratio" = petal_length / petal_width 
    •  For each species, find the average petal_ratio.'''
```

```python
df["petal_ratio"] = df["petal_length"] / df["petal_width"]
df.head()
```

```python
average_petal_ratio = df.groupby("species")["petal_ratio"].mean()

average_petal_ratio
```

```python
'''Q2 . Start by loading the TITANIC dataset.'''
```

```python
import pandas as pd

titanic_df = pd.read_csv("Titanic-Dataset.csv")
titanic_df
```

```python
'''A. Display only columns: Name, Sex, Age, Fare, Survived'''
```

```python
selected_passenger_data = titanic_df[["Name", "Sex", "Age", "Fare", "Survived"]]
selected_passenger_data
```

```python
'''B. Select passengers who: 
•  Are female 
•  Fare > 30'''
```

```python
female_high_fare = titanic_df[(titanic_df["Sex"] == "female") & (titanic_df["Fare"] > 30)]
female_high_fare
```

```python
'''C. Group by Pclass and compute: 
•  Survival rate 
•  Average Fare 
•  Average Age'''
```

```python
class_summary = titanic_df.groupby("Pclass").agg(
    survival_rate=("Survived", "mean"),
    average_fare=("Fare", "mean"),
    average_age=("Age", "mean")
)

class_summary
```
