```python
#Q1 Create a program that: 
# 1. Opens a file "names.txt" in write mode 
# 2. Writes 5 names (one per line) entered by the user 
# 3. Then opens the same file in read mode and prints all names

file = open("names.txt", "w")

for i in range(5):
    name = input("Enter a name: ")
    file.write(name + "\n")

file.close()


file = open("names.txt", "r")

print("\nNames in the file:")

for name in file:
    print(name.strip())

file.close()
```

```python
#Q2 Create a program that: 
# 1. Opens a file "log.txt" in append mode 
# 2. Adds a new log entry (like "Program run successfully") 
# 3. Opens the file in read mode and prints all logs

file = open("log.txt", "a")

log = input("Enter a log message: ")
file.write(log + "\n")

file.close()


file = open("log.txt", "r")

print("All logs:")
print(file.read())

file.close()
```

```python
#Q3 Create a program that:  
# 1. Has a list of numbers: [5, 10, 15, 20, 25] 
# 2. Uses a list comprehension to create a new list with only numbers greater than 15 
# 3. Prints the new list

numbers = [5, 10, 15, 20, 25]

greater_than_15 = [num for num in numbers if num > 15]

print(f"Numbers greater than 15: {greater_than_15}")
```

#Q4. Create a Python dictionary of 3 cities and their populations. Save it to "cities.json".
# 1. Then load the JSON and print each city and its population.
# 2. Ask the user for a new city & its population - update this info in the json file.

import json

cities = {
    "Delhi": 33800000,
    "Mumbai": 21600000,
    "Bengaluru": 14000000
}

# Save dictionary to JSON file
with open("cities.json", "w") as file:
    json.dump(cities, file)


# Load JSON file
with open("cities.json", "r") as file:
    cities = json.load(file)

# Print each city and population
print("Cities and their populations:")

for city, population in cities.items():
    print(f"{city}: {population}")

# Take new city information from user
new_city = input("\nEnter a new city: ")
new_population = int(input("Enter its population: "))

cities[new_city] = new_population


# Update the JSON file
with open("cities.json", "w") as file:
    json.dump(cities, file, indent=4)

print(f"{new_city} has been added successfully.")

```python
#Q5. Write a program that tries to open "data.txt" in read mode. If the file does not exist, catch the exception and print "File not found!".

try:
    file = open("data.txt", "r")
    print(file.read())
    file.close()

except FileNotFoundError:
    print("File not found!")
```
