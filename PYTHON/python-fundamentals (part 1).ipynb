```python
#1. Write a program that asks the user for their name and age, then prints a sentence like: Hello Asfak, you are 26 years old!

name = input("Enter your name :")
age = input("Enter your age :")

print("Hello "+ name + ", you are "+ age + " years old.")
```

```python
#2. Take two numbers as input from the user and print their sum, difference, product, and quotient.

num1 = float(input("Enter first number : "))
num2 = float(input("Enter second number : "))

sum = (num1+num2)
print("The sum is : ", sum)

difference = (num1-num2)
print("The difference is : ", difference)

multiply = (num1*num2)
print("The product is : ", multiply)

divide = (num1/num2)
print("The division is : ", divide)
```

```python
#3. Ask the user to enter two integers and one float. Convert them all to floats and print their average.

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number : "))
num3 = float(input("Enter third number :"))

avg = (num1+num2+num3)/3

print("The average of three number is : ", avg)
```

```python
'''The user enters a string containing a number (e.g., "45"). Convert it to:
• an integer
• a float
• a string again
Print all three values with their types.'''

num = input("The number entered by user: ")

int_value = int(num)
float_value = float(num)
string_value = str(num)

print(int_value, type(int_value))
print(float_value, type(float_value))
print(string_value, type(string_value))
```

```python
''' Evaluate and print the result of the following expression:
x = 10 + 3 * 2 ** 2'''

x = 10+3*2**2
print(x)
```

```python
#6. Write a program to swap values of two numbers entered by the user.

x = input("First value : ")
y = input("Second value : ")

x, y = y, x
print("After swapping- ")
print("First value: ", x)
print("Second value: ", y)
```

```python
#7. Ask the user for a temperature in Celsius (string input). Convert it to float, then calculate and print temperature in Fahrenheit.
# Conversion formula: FahrenheitTemp = (CelsiusTemp∗(9/5))+32

CelsiusTemp = input("Enter the temperature: ")
float_temp = float(CelsiusTemp)
FahrenheitTemp = (float_temp * (9/5)) + 32
print("The temperature in Fahrenheit is : ", FahrenheitTemp)
```

```python
#8. Take the radius (r) as user input and print the area.
#Use the formula: Area = π * r2 (value of π = 3.14)

r = float(input("Enter the radius of circle : "))
area = 3.14*r**2
print("The area of circle is : ", area)
```

```python
#9. Ask the user for: Principal(P), Rate(R), Time(T). Convert all to float and compute simple interest: SI =(P∗R∗T)/100

P = float(input("Enter the principal amount: "))
R = float(input("Enter the rate of interest: "))
T = float(input("Enter the time: "))
SI = (P*R*T)/100
print("The Simple Interest is: ", SI)
```

```python
'''Take a decimal number as input (like 45.78) and output its:
• integer part - 45
• fractional part - .78'''

num = input("Enter a number: ")
int_part, frac_part = num.split(".")
print("Integer part:", int_part)
print("Fractional part:", frac_part)
```
